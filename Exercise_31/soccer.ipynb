```python
wage_value = data_frame[['Name', 'Wage','Value']]
sort_wage_value = wage_value.sort_values('Wage', ascending=False)
sort_wage_value
```

```python
for column in ['Value', 'Wage']:
    parts = sort_wage_value[column].str.replace('€', '', regex=False).str.extract(
        r'([\d.]+)\s*([KMB]?)'
    )
    sort_wage_value[column] = (
        pd.to_numeric(parts[0])
        * parts[1].map({'': 1, 'K': 1_000, 'M': 1_000_000, 'B': 1_000_000_000})
    )

sort_wage_value['difference'] = sort_wage_value['Value'] - sort_wage_value['Wage']
sort_wage_value
```
